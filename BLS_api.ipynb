```python
import pandas as pd
from datetime import date 
import requests
```

```python
seriesid = "LASST060000000000003"
start_year = 2019
end_year = 2020

url =     'https://api.bls.gov/publicAPI/v1/timeseries/data/'

custom_header = {
    'Content-type': 'application/json'}


query_params = {
        'Content-type': 'application/json',
        "seriesid" : seriesid,
        "startyear" :start_year,
        "endyear" : end_year

}


bls_test = requests.get(url, data=query_params, headers=custom_header)

bls_test 

bls_test_df = pd.DataFrame(bls_test.json())
bls_test_df
```
